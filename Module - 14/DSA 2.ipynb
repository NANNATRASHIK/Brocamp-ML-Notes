```python
def bubble_sort(arr):

    n = len(arr)

    for i in range(n):

        for j in range(0, n - i - 1):

            # Compare adjacent elements
            if arr[j] > arr[j + 1]:

                # Swap them
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr


arr = [5, 3, 8, 4, 2]

print(bubble_sort(arr))
```

```python
# Selection Sort:
# Finds the smallest element from the unsorted part
# and places it at the beginning of that part.

def selection_sort(arr):

    n = len(arr)

    for i in range(n):

        # Assume the current element is the smallest
        min_index = i

        # Search for the actual smallest element
        for j in range(i + 1, n):

            if arr[j] < arr[min_index]:
                min_index = j

        # Put the smallest element at position i
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


arr = [5, 3, 8, 4, 2]

print(selection_sort(arr))
```

```python
# Insertion Sort:
# Takes one element at a time and inserts it
# into its correct position in the already sorted part.
#
# Think of arranging playing cards in your hand.

def insertion_sort(arr):

    # Start from the second element
    for i in range(1, len(arr)):

        # Element that we want to insert
        key = arr[i]

        # Index of the previous element
        j = i - 1

        # Move larger elements one position to the right
        while j >= 0 and arr[j] > key:

            arr[j + 1] = arr[j]

            j -= 1

        # Put key in its correct position
        arr[j + 1] = key

    return arr


arr = [5, 3, 8, 4, 2]

print(insertion_sort(arr))
```

```python
class Stack:
    def __init__(self):
        self.stack = []
        self.top = -1

    def push(self, values):
        self.stack.append(values)
        self.top += 1

    def pop(self):
        if self.top == -1:
            print("Stack underflow")
        else:
            value = self.stack.pop()
            self.top -= 1
            return value

    def peek(self):
        if self.top == -1:
            print("Stack is empty")
        else:
            return self.stack[self.top]

    def display(self):
        if self.top == -1:
            print("Stack is empty")
        else:
            print(self.stack)

s = Stack()

s.push(10)
s.push(20)
s.push(30)

s.display()

print("Popped:", s.pop())

print("Top:", s.peek())

s.display()
```

```python
class Queue:
    def __init__(self):
        self.queue = []
        self.front = 0
        self.rear = -1

    def enqueue(self, value):
        self.queue.append(value)
        self.rear += 1

    def dequeue(self):
        if self.front > self.rear:
            print("Queue is empty")
        else:
            value = self.queue[self.front]
            self.front += 1
            return value

    def peek(self):
        if self.front > self.rear:
            print("Queue is empty")
        else:
            return self.queue[self.front]

    def display(self):
        if self.front > self.rear:
            print("Queue is empty")
        else:
            print(self.queue[self.front:self.rear + 1])

q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()

print("Dequeued:", q.dequeue())

print("Front:", q.peek())

q.display()
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    def contains(self, value):

        current = self.root

        while current is not None:

            if value == current.value:
                return True

            elif value < current.value:
                current = current.left

            else:
                current = current.right

        return False
            
```

```python
#Max depth of BST
```

```python
def maxDepth(root):
    if root is None:
        return 0

    left = maxDepth(root.left)
    right = maxDepth(root.right)

    return max(left, right) + 1
```

```python
def isSameTree(p, q):
    if p is None and q is None:
        return True

    if p is None or q is None:
        return False

    if p.value != q.value:
        return False

    return isSameTree(p.left, q.left) and isSameTree(p.right, q.right)
```

```python
#Invert Binary Tree
```

```python
def invertTree(root):
    if root is None:
        return

    root.left, root.right = root.right, root.left

    invertTree(root.left)
    invertTree(root.right)

    return root
```

```python
def searchBST(root, val):
    while root:

        if root.val == val:
            return root
        elif root.val < val:
            root = root.left
        else:
            root = root.right

    return None
```

```python
#Symmetric Tree
```

```python
def isSymmetric(root):
    if root is None:
        return True

    def mirror(left, right):
        if left is None and right is None:
            return True

        if left is None or right is None:
            return False

        if left.val != right.val:
            return False

        return mirror(left.left, right.right) and mirror(left.right, right.left)

    return mirror(root.left, root.right)
            
```

```python
#Mini Depth
```

```python
def minDepth(root):
    if root is None:
        return

    if root.left is None:
        return minDepth(root.right) + 1

    if root.right is None:
        return minDepth(root.left) + 1
    
    left = minDepth(root.left)
    right = minDepth(root.right)

    return min(left, right) + 1
```

```python
#Count Complete Tree Nodes
```

```python
def countNode(root):
    if root is None:
        return 0

    left = countNode(root.left)
    right = countNode(root.right)

    return left + right + 1
```

```python
#Path Sum
```

```python
def pathSum(root, target):
    if root is None:
        return False

    if root.left is None and root.right is None:
        return root.val == target

    target -= root.val

    return pathSum(root.left, target) or pathSum(root.right, target)
```

```python
#Lowest Common Ancestor of a BST
```

```python
def lowestCommon(root, p, q):
    if root is None:
        return
    
    if p.val < root.val and q.val < root.val:
        return lowestCommon(root.left, p, q)
    if p.val > root.val and q.val > root.val:
        return lowestCommon(root.right, p, q)

    return root
```

```python
# Balanced Binary Tree
```

```python
def isBalanced(root):

    def height(root):
        if root is None:
            return 0
    
        return max(height(root.left), height(root.right)) + 1
    
    if root is None:
        return True

    left = height(root.left)
    right = height(root.right)

    if abs(left - right) > 1:
        return False

    return isBalanced(root.left) and isBalanced(root.right)
```

```python
def isValid(s):
    stack = []

    pairs = {
        ')':'(',
        '}':'{',
        ']':'['
    }

    for char in s:
        if char in '({[':
            stack.append(char)

        else:
            if not stack:
                return False

            if stack.pop() != pairs[char]:
                return False

    return len(stack) == 0


isValid("[")
```

```python
from collections import deque

class Stack:
    def __init__(self):
        self.q = deque()

    def push(self, x):
        self.q.append(x)

        for i in range(len(self.q) - 1):
            self.q.append(self.q.popleft())

    def pop(self):
        return self.q.popleft()

    def top(self):
        return self.q[0]

    def empty(self):
        return len(self.q) == 0
```

```python
class Queue:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def push(self, x):
        self.stack1.append(x)

    def pop(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        return self.stack2.pop()

    def peek(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        return self.stack2[-1]

    def empty(self):
        return len(self.stack1) == 0 and len(self.stack2) == 0
```

```python
def dfs(graph, node, visited):
    if node in visited:
        return

    visited.add(node)

    print(node, end=" ")

    for i in graph[node]:
        dfs(graph, i, visited)

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B'],
    'F': ['C']
}

visited = set()

dfs(graph, 'A', visited)
```

```python
def sumNode(root):
    if root is None:
        return

    if len(root.children) == 0:
        return root.val

    total = 0
    
    for child in root.children:
        total += sumNode(child)

    return total
```

```python
def countNode(root):
    if root is None:
        return 

    if len(root.children) == 0:
        return 1

    count = 0

    for child in root.children:
        count += countNode(child)

    return count
```
