```python
def isValid(s):
    stack = []

    pairs = {
        ')':'(',
        '}':'{',
        ']':'['
    }

    for char in s:
        if char in "({[":
            stack.append(char)

        else:
            if not stack:
                return False

            if stack.pop() != pairs[char]:
                return False

    return len(stack) == 0


print(isValid("[]{}"))
print(isValid("[]{"))
```

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        
        key = arr[i]

        j = i - 1

        while j >=0 and arr[j] > key:

            arr[j+1] = arr[j]
 
            j -= 1

        arr[j+1] = key

    return arr

words = ["Banana", "Apple", "Cherry", "Date"]

print(insertion_sort(words))
```

```python
class Queue:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def enque(self, x):
        self.stack1.append(x)

    def deque(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        return self.stack2.pop()

    def peek(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        return self.stack2[-1]

    def display(self):
        if not self.stack1:
            return self.stack2

        else:
            return self.stack1
    
    def empty(self):
        if len(self.stack1) == 0 and len(self.stack2) == 0:
            print("The Queue is empty")


q1 = Queue()


q1.enque(10)
q1.enque(20)
q1.enque(30)

print(q1.display())

print(q1.deque())

print(q1.display())

print(q1.empty())
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BST:
    def maxDepth(self, root):
        if root is None:
            return 0

        left = maxDepth(root.left)
        right = maxDepth(root.right)

        return max(left, right) + 1

    def insert(self, root, value):
        new_node = Node(value)
        
        if root is None:
            root = new_node

        current = root

        while current:
            if current.value < value:
                if current.left is None:
                    current.left = new_node

            else:
                if current.right is None:
                    current.right = new_node

node = Node(50)

bst = BST()

bst.insert(node,20)
bst.insert(node,30)

bst.insert(node, 40)
bst.insert(node, 55)

print(bst.maxDepth())
```
