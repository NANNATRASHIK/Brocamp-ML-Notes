```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (high + low) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] > target:
            high = mid - 1

        else:
            low = mid + 1

    return -1

arr = [10, 20, 40, 50, None]

print(binary_search(arr, 40))
```

```python
def print_numbers(n):

    if n == 0:
        return 0

    print_numbers(n - 1)
    print(n)

print_numbers(5)
```

```python
def sum_num(n):

    if n == 0:
        return 0

    return n + sum_num(n - 1)

sum_num(5)
```
