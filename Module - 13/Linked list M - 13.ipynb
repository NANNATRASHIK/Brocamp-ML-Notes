```python
class Node:
    def __init__(self, data):
        self.data = data      # Store the value/data of the node
        self.next = None      # Store the reference to the next node


def insert(head, newNode, pos):
    # If position is 1, insert the new node at the beginning
    if pos == 1:
        newNode.next = head   # New node points to the current head
        return newNode        # New node becomes the new head

    current = head           # Start from the first node

    # Move current to the node before the required position
    for i in range(pos - 2):
        if current.next is None:
            break

        current = current.next   # Move to the next node

    # Connect the new node to the next node
    newNode.next = current.next

    # Connect the current node to the new node
    current.next = newNode

    return head               # Return the head of the linked list

```

```python
def binary_search(arr, target):
    high = len(arr) - 1
    low = 0

    while low <= high:
        mid = (high + low) // 2

        if arr[mid] == target:
            return mid
        elif target > arr[mid]:
            high = mid - 1
        else:
            low = mid + 1

    return -1
```
