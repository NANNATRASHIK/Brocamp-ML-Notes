```python
s = "AAADBCA"

count = 0

dict1 = {}

for i in range(len(s) - 1):
    if s[i] not in dict1:
        dict1[s[i]] = 1

    elif s[i] in dict1:
        dict1[s[i]] += 1
 
ans = ""

for char in s:
    if char not in ans:
        ans = ans + char
        ans = ans + str(dict1[char])
        

print(ans)
```
