```python
import pandas as pd
import numpy as np

X = pd.read_csv("PCA.csv")
X = X.values

mean = np.mean(X, axis=0)
x_centered = X - mean

cov_matrix = np.cov(x_centered.T)

eigenvalue, eigenvector = np.linalg.eig(cov_matrix)

idx = np.argsort(eigenvalue)[::-1]

eigenvalue = eigenvalue[idx]
eigenvector = eigenvector[:, idx]

principal_component = eigenvector[:, :2]

x_pca = x_centered @ principal_component

print(x_pca)
```
