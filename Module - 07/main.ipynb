```python
num = int(input("Enter a number: "))

temp = num

sum = 0

while num > 0:
    digit = num % 10

    fact = 1

    for i in range(1, digit+1):
        fact *= i

    sum += fact
    
    num //= 10

if temp == sum:
    print("Strong number")
else:
    print("Not Strong number")
```

```python
num = int(input("Enter a number: "))

temp = num

power = len(str(num))

sum = 0

while num > 0:
    digit = num % 10

    sum = sum + (digit ** power)

    num //= 10

if temp==sum:
    print("Armstrong")
else:
    print("Not Armstrong")
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound():
        pass

class Cat(Animal):
    def sound(self):
        print("Meow Meow")

class Dog(Animal):
    def sound(self):
        print("Bow Bow")

cat = Cat()
dog = Dog()

cat.sound()
dog.sound()
```

```python
class BankAccount:

    def __init__(self):
        self.__acc_number = 0

    @property
    def get_acc(self):
        return self.__acc_number

    @get_acc.setter
    def get_acc(self, acc_no):
        self.__acc_number = acc_no


b1 = BankAccount()

print(b1.get_acc)

b1.get_acc = 123456789

print(b1.get_acc)
```

```python
def decorators(func):
    def wrapper():
        print("Good Morning")
        func()
        print("Good Bye")
    return wrapper

@decorators
def greet():
    print("Hello!")

greet()
```
