```python
nums = int(input("Enter a number: "))

for i in range(1, nums+1):

    prime = 1
    
    if i <= 1:
        prime = 0

    for j in range(i):
        if i % j==0:
            prime = 0
            break
            
    if prime==1:
        print(i, end=" ")
```
