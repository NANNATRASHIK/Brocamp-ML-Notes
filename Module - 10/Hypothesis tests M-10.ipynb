```python
#Z - Test
```

```python
import numpy as np

sample = [52, 49, 51, 50, 53]

pop_mean = 50
pop_std = 4

sample_mean = np.mean(sample)

n = len(sample)

z = (sample_mean - pop_mean) / (pop_std / np.sqrt(n))

print("Sample Mean:", sample_mean)
print("Z Statistic:", z)

# For right-tailed test you will use 1.645 in critical value (z > critical)
# For left-tailed test you will use -1.645 in critical value (z < critical)
# For two-tailed test you will use 1.96 in critical value (abs(z) > critical)

critical_value = 1.96

if abs(z) > critical_value:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
#T - Test
```

```python
import numpy as np

sample  = [52, 49, 51, 50, 53]

pop_mean = 50

sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

n = len(sample)

t = (sample_mean - pop_mean) / (sample_std / np.sqrt(n))

print("Sample Mean: ", sample_mean)
print("Sample Standard Deviation: ", sample_std)

critical_value = 2.776

if abs(t) > critical_value:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
#Chi-square test
```

```python
import numpy as np

observed = np.array([20, 30, 18, 32])

expected = np.array([25, 25, 25, 25])

chi_square = np.sum((observed - expected) ** 2 / expected)

print("Chi-Square Statistic:", chi_square)

critical_value = 7.815

if chi_square > critical_value:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```

```python
#ANOVA Test
```

```python
import numpy as np

# Three groups
group1 = [70, 72, 68, 71, 69]
group2 = [80, 82, 79, 81, 78]
group3 = [90, 91, 88, 89, 92]

groups = [group1, group2, group3]

# Number of groups
k = len(groups)

# Total number of observations
N = sum(len(group) for group in groups)

# Grand Mean
grand_mean = np.mean([value for group in groups for value in group])

# Sum of Squares Between (SSB)
SSB = 0

for group in groups:
    group_mean = np.mean(group)
    SSB += len(group) * (group_mean - grand_mean) ** 2

# Sum of Squares Within (SSW)
SSW = 0

for group in groups:
    group_mean = np.mean(group)

    for value in group:
        SSW += (value - group_mean) ** 2

# Degrees of Freedom
df_between = k - 1
df_within = N - k

# Mean Squares
MSB = SSB / df_between
MSW = SSW / df_within

# F Statistic
F = MSB / MSW

print("Grand Mean:", grand_mean)
print("SSB:", SSB)
print("SSW:", SSW)
print("F Statistic:", F)

# Critical value for α = 0.05, df1 = 2, df2 = 12
critical_value = 3.89

if F > critical_value:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
import numpy as np

# Sample data
data = np.array([10, 12, 13, 14, 15, 40])

# Mean
mean = np.mean(data)

# Standard deviation
std = np.std(data, ddof=1)

# Number of observations
n = len(data)

# Skewness
skewness = np.sum(((data - mean) / std) ** 3) / n

print("Skewness:", skewness)

# Interpretation
if skewness > 0:
    print("Positively Skewed")
elif skewness < 0:
    print("Negatively Skewed")
else:
    print("Symmetric")
```

```python
import numpy as np

# Sample data
data = np.array([10, 12, 13, 14, 15, 40])

# Mean
mean = np.mean(data)

# Standard deviation
std = np.std(data, ddof=1)

# Number of observations
n = len(data)

# Kurtosis
kurtosis = np.sum(((data - mean) / std) ** 4) / n

print("Kurtosis:", kurtosis)

# Interpretation
if kurtosis > 3:
    print("Leptokurtic")
elif kurtosis < 3:
    print("Platykurtic")
else:
    print("Mesokurtic")
```
