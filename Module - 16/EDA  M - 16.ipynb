Box Plot using Seaborn

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("outliers_imputation_practice.xls")

sns.boxenplot(x=df["income"])
plt.show()
```

Violinplot using Seaborn

```python
sns.violinplot(x=df["income"])
plt.show()
```

Pair Plot using Seaborn

```python
sns.pairplot(df)
plt.show()

sns.set_style("dark")
```

Heat map (Correlation - Pearson method) 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("outliers_imputation_practice.xls")

pearson_corr = df.corr(method="pearson", numeric_only=True)

sns.heatmap(pearson_corr, annot=True, cmap="coolwarm")

plt.show()
```

Heat map (Correlation - Spearman method) 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("outliers_imputation_practice.xls")

spearman_corr = df.corr(method="spearman", numeric_only=True)

sns.heatmap(spearman_corr, annot=True, cmap="coolwarm")

plt.show()
```

Heat map (Covariance)

```python
import seaborn as sns
import matplotlib.pyplot as plt

cov = df.cov(numeric_only=True)

plt.figure(figsize=(10, 6))

sns.heatmap(cov, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Covariance Matrix")
plt.show()
```
