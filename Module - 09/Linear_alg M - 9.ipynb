```python
#Matrix Multiplication 3 x 3
```

```python
A = [[20, 1, 3],
    [4, 0, 1]]

B = [[1, 6],
    [3, 4],
    [5, 6]]

result = [[0 for i in range(len(B[0]))] for i in range(len(A))]

for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            result[i][j] += A[i][k] * B[k][j]

for row in result:
    print(row)
```

```python
#Addition of Matrix 2 x 2
```

```python
A = [[1, 3],
    [4, 9]]

B = [[1, 6],
    [3, 4]]

n = len(A)

C = [[0 for i in range(n)] for i in range(n)]

for i in range(n):
    for j in range(n):
        C[i][j] = A[i][j] + B[i][j]

for row in C:
    print(row)
```

```python
#Subtraction of Matrix 2 x 2
```

```python
A = [[1, 3],
    [4, 9]]

B = [[1, 6],
    [3, 4]]

n = len(A)

C = [[0 for i in range(n)] for j in range(n)]

for i in range(n):
    for j in range(n):
        C[i][j] = A[i][j] - B[i][j]

for row in C:
    print(row)
```

```python
#Transpose of the Matrix 2 x 2
```

```python
def transpose(A):

    n = len(A)

    for i in range(n):
        for j in range(i + 1, n):
            A[i][j], A[j][i] = A[j][i], A[i][j]

    return A

A = [[1, 3],
    [4, 9]]

for row in transpose(A):
    print(row)
```

```python
#Determine of a matrix 3 x 3
```

```python
def det3(matrix):
    det = (matrix[0][0] * ((matrix[1][1] * matrix[2][2]) - (matrix[1][2] * matrix[2][1]))
        - matrix[0][1] * ((matrix[1][0] * matrix[2][2]) - (matrix[2][0] * matrix[1][2]))
        + matrix[0][2] * ((matrix[1][0] * matrix[2][1]) - (matrix[2][0] * matrix[1][1])))

    return det

A = [[1, 2 ,3],
    [4, 5, 6],
    [7, 8, 9]]

print(det3(A))

print(np.linalg.det(A))
```

```python
#Determine of matrix 2 x2
```

```python
def det2(m):
    det = (m[0][0] * m[1][1]) - (m[1][0] * m[0][1])

    return det

A = [[1, 2],
    [3, 4]]

print(det2(A))
```

```python
#Inverse of a matrix 2 x 2
```

```python
def inv2(matrix):
    det = matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]

    if det == 0:
        return "This Matrix dont have a inverse"

    return[
        [matrix[1][1]/det, -matrix[0][1]/det],
        [-matrix[1][0]/det, matrix[0][0]/det]
    ]

A = [[2, 1],
    [3, 2]]

for row in inv2(A):
    print(row)
```

```python
#Eigenvalue and Eigenvector of a matrix 2 x 2
```

```python
def eigen2(a):

    a1 = a[0][0]
    b1 = a[0][1]
    c1 = a[1][0]
    d1 = a[1][1]

    # trace = a + d
    t = a1 + d1

    # determinant = ad - bc
    det = a1 * d1 - b1 * c1

    # discriminant = trace² - 4 × determinant
    dis = t * t - 4 * det

    if dis < 0:
        print("Complex eigenvalues")
        return

    s = dis ** 0.5

    l1 = (t + s) / 2
    l2 = (t - s) / 2

    print("Eigenvalue 1 =", l1)
    print("Eigenvalue 2 =", l2)

    print()

    print("Eigenvector for", l1)

    if b1 != 0:
        x = 1
        y = (l1 - a1) / b1
    elif c1 != 0:
        y = 1
        x = (l1 - d1) / c1
    else:
        x = 1
        y = 0

    print([x, y])

    print()

    print("Eigenvector for", l2)

    if b1 != 0:
        x = 1
        y = (l2 - a1) / b1
    elif c1 != 0:
        y = 1
        x = (l2 - d1) / c1
    else:
        x = 1
        y = 0

    print([x, y])


a = [
    [4, 2],
    [1, 3]
]

eigen2(a)
```

```python
#Gradient Descend
```

```python
def gradient_descend(x, lr, itr):
    for i in range(itr):
        gradient = 2 * x
        x -= lr * gradient

    return x

print(gradient_descend(10, 0.1, 100))
```

```python
import numpy as np
import pandas as pd

# ============================
# Load Dataset
# ============================
X = pd.read_csv("PCA.csv")

# Convert DataFrame to NumPy array
X = X.values

# ============================
# Step 1: Mean Centering
# ============================
mean = np.mean(X, axis=0)
X_centered = X - mean

# ============================
# Step 2: Covariance Matrix
# ============================
cov_matrix = np.cov(X_centered.T)

# ============================
# Step 3: Eigenvalues & Eigenvectors
# ============================
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# ============================
# Step 4: Sort by Largest Eigenvalue
# ============================
idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# ============================
# Step 5: Choose Number of Components
# ============================
principal_components = eigenvectors[:, :2]

# ============================
# Step 6: Transform the Data
# ============================
X_pca = X_centered @ principal_components

# ============================
# Results
# ============================
print("Original Shape:", X.shape)
print("Reduced Shape :", X_pca.shape)

print("\nEigenvalues:")
print(eigenvalues)

print("\nPrincipal Components:")
print(principal_components)

print("\nTransformed Data:")
print(X_pca)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def gradient_decent(x, ir, itr):

    history = []

    for i in range(itr):
    
        history.append(x)
    
        gradient = 2 * x
        x -= ir * gradient

    return history

history = gradient_decent(10, 0.1, 20)

x = np.linspace(-11, 11, 400)
y = x**2
plt.plot(x, y)

plt.scatter(history, [i**2 for i in history], color="red", label="Gradient Descent")
plt.plot(history, [i**2 for i in history])

plt.show()
```
