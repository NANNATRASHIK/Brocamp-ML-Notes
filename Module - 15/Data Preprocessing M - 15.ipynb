Z - Score Detection

```python
import numpy as np

from scipy.stats import zscore

data = np.array([10, 12, 11, 13, 12, 14, 100])

z_score = zscore(data)

outliers = data[np.abs(z_score) > 2]

print(outliers)
```

IQR method

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 14, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

outlier = data[(data < lower_bond) | (data > upper_bond)]

print(outlier)
```

```python
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

outliers = df[
    (df["income"] < lower_bond) | (df["income"] > upper_bond)
]

print(outliers)
```

Remove outliers using IQR

```python
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

df_clean = df[
    (df["income"] >= lower_bond) & (df["income"] <= upper_bond)
]

print(df_clean)
```

Capping and Flooring

```python
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

df["capped_income"] = df["income"].clip(
    lower=lower_bond,
    upper=upper_bond
)

df["capped_income"].head()
```

Polynomial Features

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.impute import KNNImputer

df = pd.read_csv("outliers_imputation_practice.xls")

#PolynomialFeatures can't be used while missing values are present
imputer = KNNImputer(n_neighbors=3)

x = df[["experience_years"]]

df_imputed = pd.DataFrame(
    imputer.fit_transform(x),
    columns=["experience_years"]
)

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

df_poly = pd.DataFrame(
    poly.fit_transform(df_imputed),
    columns=["experience_years", "experience_years^2"]
)

df_poly.head(10)
```

Recursive Feature Elimination

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

X = df.drop(columns=["income"])
y = df["income"]

X = X.fillna(X.median())
y = y.fillna(y.median())

model = LinearRegression()

rfe = RFE(estimator=model, n_features_to_select=3)

rfe.fit(X, y)

selected_feature = X.columns[rfe.support_]

print(selected_feature)
```

Lasso Practical

```python
from sklearn.linear_model import Lasso
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

X = df.drop(columns=["income"])
y = df["income"]

X = X.fillna(X.median())
y = y.fillna(y.median())

lasso = Lasso(alpha=0.1)

lasso.fit(X, y)

coefficient = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": lasso.coef_
})

display(coefficient)

selected_feature = coefficient[coefficient["Coefficient"] != 0]

display(selected_feature)
```
