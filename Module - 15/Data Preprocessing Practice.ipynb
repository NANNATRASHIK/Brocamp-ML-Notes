```python
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

outliers = df[
    (df["income"] < lower_bond) | (df["income"] > upper_bond)
]

print(outliers)
```

```python
from sklearn.preprocessing import LabelEncoder, PolynomialFeatures
from sklearn.impute import KNNImputer
import pandas as pd
import numpy as np

df = pd.read_csv("outliers_imputation_practice.xls")

df["experience_level"] = pd.cut(
    df["experience_years"],
    bins=[0, 5, 15, 100],
    labels=["Junior", "Mid", "Senior"]
)

df["log_income"] = np.log1p(df["income"])

df["project_per_hour"] = df["projects_completed"] / df["hours_per_week"].replace(0, pd.NA)

imputer = KNNImputer(n_neighbors=3)

poly = PolynomialFeatures(degree=2, include_bias=False)

incoder = LabelEncoder()

df["experience_years"] = imputer.fit_transform(df[["experience_years"]])

df_poly = pd.DataFrame(
    poly.fit_transform(df[["experience_years"]]),
    columns=["experience_years", "experience_years^2"]
    )

df_enoded = pd.get_dummies(
    df,
    columns=["experience_level"],
    dtype=int
)

df["experience_level_encoded"] = incoder.fit_transform(df["experience_level"])

display(df.head())

display(df_poly.head())

display(df_enoded.head())
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer, KNNImputer
from sklearn.preprocessing import MinMaxScaler, StandardScaler, PolynomialFeatures, LabelEncoder, RobustScaler
from scipy.stats import zscore
import pandas as pd 
import numpy as np

df = pd.read_csv("outliers_imputation_practice.xls")

kimputer = KNNImputer(n_neighbors=3)
mimputer = IterativeImputer(max_iter=10, random_state=53)
simputer = SimpleImputer(strategy="mean")
poly = PolynomialFeatures(degree=2, include_bias=False)

df = pd.read_csv("outliers_imputation_practice.xls")

print("Imputation")

df_kimputed = pd.DataFrame(
    kimputer.fit_transform(df),
    columns=df.columns
)

df_mimputed = pd.DataFrame(
    mimputer.fit_transform(df),
    columns=df.columns
)

df_simputed = pd.DataFrame(
    simputer.fit_transform(df),
    columns=df.columns
)

display(df_kimputed.head())
display(df_mimputed.head())
display(df_simputed.head())

print("Normalization")

mscaler = MinMaxScaler()
zscaler = StandardScaler()
rscaler = RobustScaler()

df_mscaled = pd.DataFrame(
    mscaler.fit_transform(df),
    columns=df.columns
)

df_zscaled = pd.DataFrame(
    zscaler.fit_transform(df),
    columns=df.columns
)

df_rscaled = pd.DataFrame(
    rscaler.fit_transform(df),
    columns=df.columns
)

print("Transformation")

df_poly = pd.DataFrame(
    poly.fit_transform(df_mimputed[["experience_years"]]),
    columns=["experience_years", "experience_years ^ 2"]
)

display(df_poly)

display(df_mscaled.head())
display(df_zscaled.head())
display(df_rscaled.head())
```

```python
from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import KNNImputer
import pandas as pd

df = pd.read_csv("outliers_imputation_practice.xls")

correlation = df.corr(numeric_only=True)

display(correlation)

correlation = df.corr(numeric_only=True)["income"].sort_values(ascending=False)

display(correlation)

imputer = KNNImputer(n_neighbors=4)

df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

selector = VarianceThreshold(threshold=3)

selector.fit(df_imputed)

selected = df_imputed.columns[selector.get_support()]

print(selected)
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

X = df.drop(columns=["income"])
y = df["income"]

X = X.fillna(X.median())
y = y.fillna(y.median())

model = LinearRegression()

rfe = RFE(estimator=model, n_features_to_select=3)

rfe.fit(X, y)

selected = X.columns[rfe.support_]

print(selected)
```
