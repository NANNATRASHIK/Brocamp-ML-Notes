```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]

y = [2, 4, 6, 8, 10]

plt.plot(x,y)

plt.title("Simple line")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
days = [1,2,3,4,5]
sales = [100,120,150,170,200]

plt.plot(days, sales)

plt.show()
```

```python
plt.plot(
    x,
    y,
    color="green",
    linestyle="--",
    linewidth=2,
    marker="*"
)
```

```python
height = [160,165,170,175,180]
weight = [55,60,68,75,82]

plt.scatter(height, weight)

plt.show()
```

```python
plt.scatter(
    x, 
    y,
    c = "black",
    s = 50,
    marker = "*"
)
```

```python
students = ["A","B","C","D"]
marks = [90,75,88,95]

plt.bar(students, marks)
```

```python
ages = [18,18,19,19,20,20,21,22,22,23]

plt.hist(ages)

plt.show()
```

```python
plt.hist(
    ages,
    bins = 10,
    edgecolor="black"
)
```

```python
languages = ["Python","Java","C++"]
students = [50,30,20]

plt.pie(
    students,
    labels = languages
)
```

```python
salary = [20,22,25,26,27,28,30,90]

plt.boxplot(salary)

plt.show()
```
